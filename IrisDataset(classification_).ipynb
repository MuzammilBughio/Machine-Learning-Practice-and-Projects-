```python
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris(as_frame=True)
df = iris.frame
```

```python
df.head()
```

```python
X = df.drop('target' , axis =1)
y = df['target']
```

```python
# Logistic Regression ... without scaler

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix , classification_report , accuracy_score , precision_score , recall_score

X_train , X_test , y_train , y_test = train_test_split(X , y , train_size=0.2 , random_state=42 , stratify=y  ) # stratify used for balanced splitting

model = LogisticRegression()
model.fit(X_train , y_train)
y_pred = model.predict(X_test)

print("Confusion Matrix : \n" , confusion_matrix(y_test , y_pred))
print("Classification Report :\n " , classification_report(y_test , y_pred))
print("Acuuracy  : " , accuracy_score(y_test , y_pred))
print("Precision : " , precision_score(y_test , y_pred , average='macro' ))
print("Recall  : " , recall_score(y_test , y_pred , average='macro'))
```

```python
# Logistic Regression ... with scaler

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix , classification_report , accuracy_score , precision_score , recall_score
from sklearn.preprocessing import StandardScaler

X_train , X_test , y_train , y_test = train_test_split(X , y , train_size=0.2 , random_state=42 , stratify=y  ) # stratify used for balanced splitting

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = LogisticRegression()
model.fit(X_train_scaled , y_train)
y_pred = model.predict(X_test_scaled)

print("Confusion Matrix : \n" , confusion_matrix(y_test , y_pred))
print("Classification Report :\n " , classification_report(y_test , y_pred))
print("Acuuracy  : " , accuracy_score(y_test , y_pred))
print("Precision : " , precision_score(y_test , y_pred , average='macro' ))
print("Recall  : " , recall_score(y_test , y_pred , average='macro'))
```

```python
# Naive Bayes

from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import confusion_matrix , classification_report , accuracy_score , precision_score , recall_score

X_train , X_test , y_train , y_test = train_test_split(X , y , train_size=0.2 , random_state=42 , stratify=y  ) # stratify used for balanced splitting

model = GaussianNB()
model.fit(X_train , y_train)
y_pred = model.predict(X_test)

print("Confusion Matrix : \n" , confusion_matrix(y_test , y_pred))
print("Classification Report :\n " , classification_report(y_test , y_pred , target_names=iris.target_names))
print("Acuuracy  : " , accuracy_score(y_test , y_pred))
print("Precision : " , precision_score(y_test , y_pred , average='macro' ))
print("Recall  : " , recall_score(y_test , y_pred , average='macro'))
```

```python
# KNN ... without scaler and without (CV and Pipeline)

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix , classification_report , accuracy_score , precision_score , recall_score

X_train , X_test , y_train , y_test = train_test_split(X , y , train_size=0.2 , random_state=42 , stratify=y  ) # stratify used for balanced splitting

model = KNeighborsClassifier()
model.fit(X_train , y_train)
y_pred = model.predict(X_test)

print("Confusion Matrix : \n" , confusion_matrix(y_test , y_pred))
print("Classification Report :\n " , classification_report(y_test , y_pred))
print("Acuuracy  : " , accuracy_score(y_test , y_pred))
print("Precision : " , precision_score(y_test , y_pred , average='macro' ))
print("Recall  : " , recall_score(y_test , y_pred , average='macro'))
```

```python
# KNN ... with scaler but without (CV and Pipeline)

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix , classification_report , accuracy_score , precision_score , recall_score
from sklearn.preprocessing import StandardScaler

X_train , X_test , y_train , y_test = train_test_split(X , y , train_size=0.2 , random_state=42 , stratify=y  ) # stratify used for balanced splitting

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

model = KNeighborsClassifier(n_neighbors=7)
model.fit(X_train_scaled , y_train)
y_pred = model.predict(X_test_scaled)


print("Confusion Matrix : \n" , confusion_matrix(y_test , y_pred))
print("Classification Report :\n " , classification_report(y_test , y_pred))
print("Acuuracy  : " , accuracy_score(y_test , y_pred))
print("Precision : " , precision_score(y_test , y_pred , average='macro' ))
print("Recall  : " , recall_score(y_test , y_pred , average='macro'))
```

```python
# KNN ... with scaler and with (CV and Pipeline)

from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import confusion_matrix , classification_report , accuracy_score , precision_score , recall_score
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import GridSearchCV
from sklearn.pipeline import Pipeline

X_train , X_test , y_train , y_test = train_test_split(X , y , train_size=0.2 , random_state=42 , stratify=y  ) # stratify used for balanced splitting
scaler = StandardScaler()
knn_model = KNeighborsClassifier()
pipeline = Pipeline([
    ('scaler' , scaler),
    ('knn' , knn_model)
])

param_grid = {"knn__n_neighbors" : [ 3 , 5 , 7 , 9 , 11]}
model_KNNCV = GridSearchCV(pipeline , param_grid , cv = 5 )


model_KNNCV.fit(X_train, y_train)
y_pred = model_KNNCV.predict(X_test)


print("Confusion Matrix : \n" , confusion_matrix(y_test , y_pred))
print("Classification Report :\n " , classification_report(y_test , y_pred))
print("Acuuracy  : " , accuracy_score(y_test , y_pred))
print("Precision : " , precision_score(y_test , y_pred , average='macro' ))
print("Recall  : " , recall_score(y_test , y_pred , average='macro'))
```
